# Import Library

```python
import numpy as np
from matplotlib import pyplot as plt

%matplotlib inline
```

# Percobaan 1 - Line Plot

*Line plot* (grafik garis) digunakan untuk menyajikan visualisasi data runtun waktu atau hubungan kontinu antara dua variabel (dalam hal ini, waktu terhadap amplitudo). Garis kontinu terbentuk dengan menghubungkan titik-titik koordinat $(x, y)$ yang berurutan.

```python
# Data for plotting
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

plt.figure()
plt.plot(t, s)
plt.title('Plot Nilai Fungsi 2*Sin(t1)')
plt.xlabel('Waktu (s)')
plt.ylabel('Amplitudo (mV)')
plt.grid()
plt.show()
```

**Penjelasan Kode:**
* `np.arange(0.0, 2.0, 0.01)`: Membuat deret angka dari `0.0` hingga `2.0` dengan interval `0.01` sebagai sumbu $x$ (waktu).
* `s = 1 + np.sin(2 * np.pi * t)`: Menghitung nilai fungsi sinusoidal sebagai sumbu $y$ (amplitudo).
* `plt.figure()`: Menginisialisasi *figure* atau kanvas plot baru dengan ukuran default.
* `plt.plot(t, s)`: Menggambar grafik garis dengan data `t` pada sumbu $x$ dan `s` pada sumbu $y$.
* `plt.title()`, `plt.xlabel()`, `plt.ylabel()`: Memberikan judul grafik serta label nama pada sumbu horizontal dan vertikal.
* `plt.grid()`: Menampilkan garis bantu kisi-kisi (grid) untuk mempermudah pembacaan nilai koordinat.
* `plt.show()`: Merender dan menampilkan visualisasi akhir ke layar.

# Percobaan 2 - Line Plot dengan 'figsize'

Secara teori, ukuran kanvas visualisasi (*aspect ratio*) sangat menentukan keterbacaan data. Ukuran yang terlalu kecil dapat membuat kurva yang padat menjadi sulit dianalisis, sedangkan ukuran yang tepat memberikan ruang yang proporsional bagi label dan detail kurva.

```python
# Data for plotting
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)

plt.figure(figsize=(10,5))
plt.plot(t, s)
plt.title('Plot Nilai Fungsi 2*Sin(t1)')
plt.xlabel('Waktu (s)')
plt.ylabel('Amplitudo (mV)')
plt.grid()
plt.show()
```

**Penjelasan Kode:**
* Perbedaan utama dari Percobaan 1 terletak pada baris `plt.figure(figsize=(10,5))`.
* Argumen `figsize=(10,5)` menentukan dimensi lebar kanvas sebesar 10 inci dan tinggi sebesar 5 inci. Hal ini membuat visualisasi grafik menjadi lebih melebar secara horizontal dibandingkan ukuran standar.

# Percobaan 3 - Bar Chart (Vertical)

*Bar Chart* (Diagram Batang) Vertikal digunakan untuk membandingkan nilai atau frekuensi antar kategori diskret secara visual. Tinggi dari setiap batang merepresentasikan kuantitas dari kategori tersebut.

```python
langs = ['C', 'C++', 'Java', 'Python', 'PHP']
students = [23,17,35,29,12]

plt.figure(figsize=(10,5))
plt.bar(langs,students)
plt.ylabel('Pengguna')
plt.xlabel('Bahasa Pemrograman')
plt.show()
```

**Penjelasan Kode:**
* `langs`: Berisi daftar kategori berupa nama bahasa pemrograman (data nominal).
* `students`: Berisi nilai numerik yang mewakili jumlah pengguna masing-masing bahasa pemrograman.
* `plt.bar(langs, students)`: Fungsi khusus untuk membuat diagram batang vertikal, di mana argumen pertama (`langs`) menjadi sumbu $x$ dan argumen kedua (`students`) menjadi tinggi batang pada sumbu $y$.

# Percobaan 4 - Bar Chart (Horizontal)

```python
y_pos = np.arange(len(langs))

plt.figure(figsize=(10,5))
plt.barh(y_pos, students)
plt.yticks(y_pos, langs)
plt.xlabel('Pengguna')
plt.ylabel('Bahasa Pemrograman')
plt.title('Programming language usage')
plt.show()
```

### Penjelasan Percobaan 4

**Teori Dasar:**
*Horizontal Bar Chart* sangat berguna ketika label kategori memiliki nama yang cukup panjang, sehingga jika digambar secara vertikal label tersebut akan saling bertumpuk. Panjang batang ke arah kanan merepresentasikan kuantitas data.

**Penjelasan Kode:**
* `y_pos = np.arange(len(langs))`: Membuat indeks numerik (0 sampai 4) sebagai posisi penempatan batang pada sumbu $y$.
* `plt.barh(y_pos, students)`: Menggambar diagram batang secara horizontal (`barh`), dengan `y_pos` sebagai posisi vertikal dan `students` sebagai panjang batang horizontalnya.
* `plt.yticks(y_pos, langs)`: Mengubah skala indeks numerik pada sumbu $y$ menjadi label teks nama bahasa pemrograman yang sesuai.

# Percobaan 5 - Pie Chart

```python
plt.figure(figsize=(10,5))
plt.pie(students, labels = langs,autopct='%1.2f%%')
plt.show()
```

### Penjelasan Percobaan 5

**Teori Dasar:**
*Pie Chart* (Diagram Lingkaran) digunakan untuk menunjukkan kontribusi proporsi atau persentase masing-masing kategori terhadap keseluruhan (total 100%).

**Penjelasan Kode:**
* `plt.pie(students, labels = langs, autopct='%1.2f%%')`: Menggambar diagram lingkaran.
    * `students` menentukan ukuran sudut setiap irisan lingkaran.
    * `labels = langs` memberikan label teks di luar setiap irisan.
    * `autopct='%1.2f%%'` memformat tampilan nilai persentase di dalam irisan lingkaran secara otomatis dengan 2 angka di belakang desimal.

# Percobaan 6 - Pie Chart dengan 'explode'

```python
explode = (0, 0.1, 0, 0, 0)

plt.figure(figsize=(10,5))
plt.pie(students, explode=explode, labels = langs,autopct='%1.2f%%')
plt.show()
```

### Penjelasan Percobaan 6

**Teori Dasar:**
Fitur *Explode* pada diagram lingkaran digunakan untuk memberikan penekanan (*highlight*) pada salah satu atau beberapa irisan tertentu dengan cara memisahkan atau menarik irisan tersebut keluar dari pusat lingkaran.

**Penjelasan Kode:**
* `explode = (0, 0.1, 0, 0, 0)`: Sebuah *tuple* yang menentukan jarak pergeseran tiap irisan dari pusat lingkaran. Nilai `0.1` pada indeks kedua berarti irisan untuk kategori 'C++' akan ditarik keluar sebesar 10% dari radius lingkaran, sementara irisan lainnya tetap berada di pusat (`0`).

# Percobaan 7 - 2 Subplot

```python
import numpy as np
import matplotlib.pyplot as plt

x1 = np.linspace(0.0, 5.0)
x2 = np.linspace(0.0, 2.0)

y1 = np.cos(2 * np.pi * x1) * np.exp(-x1)
y2 = np.cos(2 * np.pi * x2)

plt.figure(figsize=(10,5))
plt.subplot(2, 1, 1)
plt.plot(x1, y1)
plt.title('Subplot ke 1')
plt.ylabel('Amplitudo (mV)')
plt.xlabel('time (s)')
plt.grid()

plt.subplot(2, 1, 2)
plt.plot(x2, y2)
plt.title('Subplot ke 2')
plt.xlabel('time (ms)')
plt.ylabel('Amplitudo (mV)')
plt.tight_layout()
plt.show()
```

### Penjelasan Percobaan 7

**Teori Dasar:**
*Subplot* digunakan untuk menampilkan beberapa grafik berbeda dalam satu kanvas (*figure*) yang sama. Konsep koordinat kisi (*grid*) digunakan untuk menentukan posisi baris dan kolom dari grafik tersebut.

**Penjelasan Kode:**
* `plt.subplot(2, 1, 1)`: Menyiapkan area plot dengan membagi kanvas menjadi kisi ukuran **2 baris dan 1 kolom**, lalu mengaktifkan area pertama (**baris 1**) untuk digambar.
* `plt.subplot(2, 1, 2)`: Mengaktifkan area kedua (**baris 2**) pada kisi yang sama untuk menggambar grafik berikutnya.
* `plt.tight_layout()`: Menata layout secara otomatis agar elemen antar-subplot (seperti judul dan label sumbu) tidak saling tumpang tindih.

# Percobaan 8 - 4 Subplot

```python
plt.figure(figsize=(10,10))
plt.subplot(2, 2, 1)
plt.pie(students, labels = langs,autopct='%1.2f%%')
plt.title('Subplot ke 1')

plt.subplot(2, 2, 2)
explode = (0, 0.1, 0, 0, 0)
plt.pie(students, explode=explode, labels = langs,autopct='%1.2f%%')
plt.title('Subplot ke 2')

plt.subplot(2, 2, 3)
plt.bar(langs,students)
plt.ylabel('Pengguna')
plt.xlabel('Bahasa Pemrograman')
plt.title('Subplot ke 3')

plt.subplot(2, 2, 4)
plt.barh(y_pos, students)
plt.yticks(y_pos, langs)
plt.xlabel('Pengguna')
plt.ylabel('Bahasa Pemrograman')
plt.title('Subplot ke 4')

plt.tight_layout()
plt.show()
```

### Penjelasan Percobaan 8

**Teori Dasar:**
Pengembangan dari konsep subplot sebelumnya, di mana kita dapat mengombinasikan berbagai jenis visualisasi yang berbeda (*pie chart*, *bar chart*) ke dalam satu bingkai grid multi-kolom dan multi-baris (matriks $2 \times 2$).

**Penjelasan Kode:**
* `plt.subplot(2, 2, n)`: Membagi kanvas menjadi matriks **2 baris dan 2 kolom** (total 4 area plot):
    * `plt.subplot(2, 2, 1)`: Posisi kiri atas (diisi *Pie Chart* standar).
    * `plt.subplot(2, 2, 2)`: Posisi kanan atas (diisi *Pie Chart* dengan efek explode).
    * `plt.subplot(2, 2, 3)`: Posisi kiri bawah (diisi *Vertical Bar Chart*).
    * `plt.subplot(2, 2, 4)`: Posisi kanan bawah (diisi *Horizontal Bar Chart*).
* `plt.tight_layout()` menyusun keseluruhan 4 subplot tersebut agar rapi dan tidak saling bertabrakan.
