Citra adalah suatu representasi atau gambaran yang menyerupai, mirip, atau tiruan dari suatu objek. Pada umumnya citra terbagi menjadi 2, yaitu citra yang bersifat analog dan citra yang bersifat digital.

## 1. Citra Analog
Citra analog adalah citra yang bersifat berkelanjutan (kontinu), seperti:
- Gambar pada monitor televisi
- Foto sinar X
- Foto yang tercetak di kertas foto
- Lukisan & pemandangan alam
- Hasil CT scan
- Gambar-gambar yang terekam pada pita kaset, dan lain sebagainya.

Citra analog tidak dapat direpresentasikan dalam komputer secara langsung sehingga tidak bisa diproses di komputer tanpa melalui proses konversi analog ke digital (*Analog-to-Digital Conversion*).

Citra analog dihasilkan dari alat-alat analog seperti video kamera analog, kamera foto analog, CT scan, sensor rontgen untuk foto thorax, sensor gelombang pendek pada sistem radar, sensor ultrasound pada sistem USG, dan lain-lain.

## 2. Citra Digital
Citra digital merupakan representasi dari citra yang diambil oleh mesin dengan bentuk pendekatan berdasarkan **sampling** dan **kuantisasi**.
- **Sampling**: Menyatakan besarnya kotak-kotak yang disusun dalam baris dan kolom. Dengan kata lain, sampling pada citra menyatakan besar kecilnya ukuran pixel (titik) pada citra.
- **Kuantisasi**: Menyatakan besarnya nilai tingkat kecerahan yang dinyatakan dalam nilai tingkat keabuan (*grayscale*) sesuai dengan jumlah bit biner yang digunakan oleh mesin, atau menyatakan jumlah warna yang ada pada citra.

### 2.2. Citra Berskala Keabuan (Grayscale)
Citra grayscale menangani gradasi warna hitam dan putih yang menghasilkan efek warna abu-abu. Nilai intensitas berkisar antara **0 sampai dengan 255**:
- Nilai **0** menyatakan hitam sempurna.
- Nilai **255** menyatakan putih sempurna.
- Nilai **1–254**: berbagai tingkat keabuan.

![Nama Gambar](https://catatanpeneliti.wordpress.com/wp-content/uploads/2013/06/gmbr11.jpg)

```python
# @title Kode berikut membaca citra berwarna, mengonversinya menjadi grayscale, menampilkan kedua citra, dan mencetak nilai pikselnya.
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Membaca citra asli dan mengonversinya menjadi grayscale
img = Image.open('kucing.jpg')
gray_img = img.convert('L')

# Mengubah citra menjadi array NumPy
img_rgb = np.array(img.convert('RGB'))
gray_array = np.array(gray_img)

# Menampilkan dimensi citra
print("Dimensi citra RGB:", img_rgb.shape)
print("Dimensi citra Grayscale:", gray_array.shape)

# Menampilkan nilai piksel grayscale pada 3x3 piksel pertama
print("\nMatriks nilai piksel Grayscale (3x3 pertama):")
print(gray_array[:3, :3])

# Membuat dua tampilan berdampingan
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Menampilkan citra asli RGB
axes[0].imshow(img_rgb)
axes[0].set_title("Citra Asli (RGB)")
axes[0].axis('off')

# Menampilkan citra grayscale
axes[1].imshow(gray_array, cmap='gray', vmin=0, vmax=255)
axes[1].set_title("Citra Grayscale")
axes[1].axis('off')

plt.tight_layout()
plt.show()
```

Misalnya, satu piksel pada citra RGB memiliki nilai `[200, 100, 50]`. Setelah dikonversi menjadi grayscale, piksel tersebut memiliki satu nilai intensitas, sekitar `124`.

Secara umum, konversi grayscale menggunakan pembobotan kanal warna:

\\[ Y = 0{,}299R + 0{,}587G + 0{,}114B \\]

Rumus tersebut menunjukkan bahwa kanal hijau memiliki kontribusi terbesar terhadap intensitas keabuan, diikuti merah dan biru.

Perbedaan dimensi array:

- RGB: `(tinggi, lebar, 3)` karena setiap piksel memiliki tiga kanal warna.
- Grayscale: `(tinggi, lebar)` karena setiap piksel hanya memiliki satu nilai intensitas.

> Grayscale menyederhanakan representasi citra dari tiga kanal warna menjadi satu kanal intensitas. Hal ini sering berguna untuk analisis tepi, segmentasi, penghitungan histogram, dan berbagai proses pengolahan citra lainnya.

### 2.3 Citra Biner
Citra biner adalah citra dengan setiap piksel hanya dinyatakan dengan sebuah nilai dari dua kemungkinan, yaitu **0 dan 1** (kedalaman bit 1-bit):
- Nilai **0** menyatakan warna hitam.
- Nilai **1** menyatakan warna putih.
Citra biner banyak dipakai dalam pemrosesan citra, misalnya untuk kepentingan memperoleh tepi bentuk suatu objek (*edge detection*).

![Nama Gambar](https://encrypted-tbn0.gstatic.com/images?q=tbn:ANd9GcSlB1q616Jss3bTHkpyKZwfEjpJ6SKix1o3gOr4zEQ87B2YW2B3UyyKZ9UU&s=10)

```python
# @title Kode berikut membaca gambar, mengubahnya menjadi grayscale, kemudian mengubahnya menjadi citra biner menggunakan metode thresholding.

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Membaca citra asli
img = Image.open('kucing.jpg').convert('RGB')

# Mengubah citra RGB menjadi grayscale
gray_img = img.convert('L')

# Mengubah citra grayscale menjadi array NumPy
gray_array = np.array(gray_img)

# Menentukan nilai ambang (threshold)
threshold = 128

# Melakukan thresholding untuk menghasilkan citra biner
# Piksel >= 128 menjadi putih (255)
# Piksel < 128 menjadi hitam (0)
binary_array = np.where(gray_array >= threshold, 255, 0).astype(np.uint8)

# Menampilkan dimensi citra
print("Dimensi citra RGB:", np.array(img).shape)
print("Dimensi citra Grayscale:", gray_array.shape)
print("Dimensi citra Biner:", binary_array.shape)

# Menampilkan nilai piksel pada area 3x3 pertama
print("\nNilai piksel Grayscale (3x3):")
print(gray_array[:3, :3])

print("\nNilai piksel Biner (3x3):")
print(binary_array[:3, :3])

# Membuat visualisasi citra
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Menampilkan citra asli
axes[0].imshow(img)
axes[0].set_title("Citra RGB")

# Menampilkan citra grayscale
axes[1].imshow(gray_array, cmap='gray', vmin=0, vmax=255)
axes[1].set_title("Citra Grayscale")

# Menampilkan citra biner
axes[2].imshow(binary_array, cmap='gray', vmin=0, vmax=255)
axes[2].set_title("Citra Biner")

# Menghilangkan sumbu koordinat
for ax in axes:
    ax.axis('off')

plt.tight_layout()
plt.show()
```

#### **Penjelasan proses thresholding**

Thresholding adalah proses mengelompokkan piksel berdasarkan nilai ambang tertentu.

Misalnya, ditentukan nilai ambang \\(T=128\\). Aturan konversinya adalah:

\\[ B(x,y)= \begin{cases} 255, & I(x,y)\geq128\\\ 0, & I(x,y)<128 \end{cases} \\]

Keterangan:

- \\(I(x,y)\\) adalah nilai intensitas piksel grayscale.
- \\(T\\) adalah nilai ambang (threshold).
- \\(B(x,y)\\) adalah nilai piksel hasil citra biner.

Contoh perubahan nilai piksel:

| Nilai grayscale | Nilai biner | Hasil |
| --------------- | ----------- | ----- |
| 35              | 0           | Hitam |
| 90              | 0           | Hitam |
| 127             | 0           | Hitam |
| 128             | 255         | Putih |
| 180             | 255         | Putih |
| 240             | 255         | Putih |

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Membaca citra berwarna
img = cv2.imread('kucing.jpg')

if img is not None:

    # Mengubah citra BGR menjadi RGB agar warna tampil benar
    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Mengubah citra berwarna menjadi grayscale
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

    # Menentukan nilai ambang (threshold)
    threshold = 128

    # Melakukan thresholding
    # Piksel >= 128 menjadi putih (255)
    # Piksel < 128 menjadi hitam (0)
    _, binary = cv2.threshold(
        gray, threshold, 255, cv2.THRESH_BINARY
    )

    # Menampilkan nilai piksel
    print("Matriks Grayscale (3x3 piksel pertama):")
    print(gray[:3, :3])

    print("\nMatriks Biner (3x3 piksel pertama):")
    print(binary[:3, :3])

    # Menampilkan tiga citra berdampingan
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))

    axes[0].imshow(img_rgb)
    axes[0].set_title("Citra Asli (RGB)")

    axes[1].imshow(gray, cmap='gray', vmin=0, vmax=255)
    axes[1].set_title("Citra Grayscale")

    axes[2].imshow(binary, cmap='gray', vmin=0, vmax=255)
    axes[2].set_title("Hasil Thresholding (T=128)")

    for ax in axes:
        ax.axis('off')

    plt.tight_layout()
    plt.show()

else:
    print("File kucing.jpg tidak ditemukan.")
```
