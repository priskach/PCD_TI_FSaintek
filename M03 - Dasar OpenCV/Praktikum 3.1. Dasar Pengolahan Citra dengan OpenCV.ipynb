Scikit-Image adalah library pemrosesan gambar berbasis Python yang memiliki beberapa bagian ditulis dalam Cython (superset Python yang dirancang memiliki kinerja seperti bahasa C). Scikit-image menggunakan array NumPy sebagai objek gambar.

### C. SciPy
SciPy digunakan untuk perhitungan matematis dan ilmiah, serta dapat melakukan pemrosesan gambar multi-dimensi menggunakan submodul `scipy.ndimage` yang beroperasi pada array NumPy N-dimensi.

**Operasi umum:** Reading Images, Image Segmentation, Convolution, Face Detection, Feature Extraction.

**Contoh Kode: Pemrosesan Gambar Multi-dimensi (Gaussian Blur)**

````
from scipy import misc, ndimage
from scipy.datasets import face
from matplotlib import pyplot as plt

# Memuat contoh gambar wajah dari modul misc SciPy
face = face()

# Menerapkan Gaussian filter dengan sigma = 3 untuk pemburaman multi-dimensi
blurred_face = ndimage.gaussian_filter(face, sigma=3)

# Menampilkan hasil menggunakan Matplotlib
plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.imshow(face)
plt.title("Citra Asli (Original)")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(blurred_face)
plt.title("Hasil Gaussian Blur (sigma=3)")
plt.axis('off')

plt.show()
````

### D. Pillow / PIL
PIL (Python Imaging Library) adalah open-source library untuk tugas pemrosesan gambar seperti membaca, mengubah skala (*scaling*), dan menyimpan dalam format berbeda.

### Latihan 1: 'CN' Color Selection pada Matplotlib

Matplotlib mengonversi warna "CN" ke RGBA saat
menggambar. Bagian Styling berisi informasi tambahan
tentang mengontrol warna dan properti gaya.


````
import matplotlib as mpl
import matplotlib.pyplot as plt
import numpy as np

th = np.linspace(0, 2 * np.pi, 128)


def demo(sty):
  mpl.style.use(sty)
  fig, ax = plt.subplots(figsize=(3, 3))
  ax.set_title("style: {!r}".format(sty), color="C0")
  ax.plot(th, np.cos(th), "C1", label="C1")
  ax.plot(th, np.sin(th), "C2", label="C2")
  ax.legend()
  plt.show()


demo("default")
demo("seaborn-v0_8" if "seaborn-v0_8" in plt.style.available else "default")
````

### Latihan 2: Perbandingan Warna X11/CSS4 dan xkcd


Perbandingan warna X11/CSS4 dan XKCD biasanya dibahas dalam konteks Python, khususnya pustaka `Matplotlib` untuk visualisasi data.

Keduanya merupakan kumpulan nama warna (named colors) yang dapat digunakan untuk menentukan warna grafik, garis, batang, titik, maupun elemen visual lainnya. Perbedaan utamanya terletak pada asal penamaan, jumlah warna, dan karakteristik nama warnanya.

## 1. Perbandingan X11/CSS4 dan XKCD

| Aspek                      | X11/CSS4                                        | XKCD                                                |
| -------------------------- | ----------------------------------------------- | --------------------------------------------------- |
| Asal                       | Warna dari sistem X11 dan standar warna CSS     | Survei penamaan warna oleh XKCD                     |
| Karakter nama              | Nama warna standar dan umum                     | Nama warna berdasarkan istilah sehari-hari          |
| Contoh nama                | `red`, `blue`, `darkgreen`, `coral`             | `sky blue`, `dusty purple`, `pea green`             |
| Jumlah warna di Matplotlib | CSS4: 148 nama warna                            | XKCD: 949 nama warna                                |
| Penggunaan                 | Grafik formal, laporan, dan visualisasi standar | Visualisasi dengan variasi warna yang lebih beragam |
| Format pemanggilan         | `colors.CSS4_COLORS`                            | `colors.XKCD_COLORS`                                |

Catatan: X11 dan CSS4 bukan kumpulan yang sepenuhnya identik. Matplotlib menyediakan daftar warna X11 sebagai bagian dari referensi warna bernama, sedangkan `CSS4_COLORS` secara khusus mengacu pada daftar warna CSS4.

## 2. Contoh perbedaan warna

image.png

Menurut dokumentasi Matplotlib, terdapat 95 nama warna yang sama pada kedua daftar tersebut, tetapi hampir semuanya memiliki nilai warna berbeda. Hanya `black`, `white`, dan `cyan` yang memiliki nilai identik di kedua daftar.&#x20;

[image](https://www.google.com/s2/favicons?domain=https://en.matplotlib.net\&sz=32)

Kode tersebut menampilkan jumlah warna yang tersedia pada versi Matplotlib yang digunakan.

X11/CSS4 lebih sesuai untuk penggunaan warna standar dan konsisten, sedangkan XKCD menyediakan pilihan warna yang jauh lebih banyak dengan nama yang lebih deskriptif dan beragam. Keduanya dapat digunakan dalam visualisasi data menggunakan Matplotlib.



````
import matplotlib._color_data as mcd
import matplotlib.patches as mpatch
import matplotlib.pyplot as plt

overlap = {
    name for name in mcd.CSS4_COLORS if "xkcd:" + name in mcd.XKCD_COLORS
}
fig = plt.figure(figsize=[9, 5])
ax = fig.add_axes([0, 0, 1, 1])
n_groups = 3
n_rows = len(overlap) // n_groups + 1

for j, color_name in enumerate(sorted(overlap)):
  css4 = mcd.CSS4_COLORS[color_name]
  xkcd = mcd.XKCD_COLORS["xkcd:" + color_name].upper()
  col_shift = (j // n_rows) * 3
  y_pos = j % n_rows
  text_args = dict(
      va="center",
      fontsize=10,
      weight="bold" if css4 == xkcd else None,
  )
  ax.add_patch(mpatch.Rectangle((0 + col_shift, y_pos), 1, 1, color=css4))
  ax.add_patch(mpatch.Rectangle((1 + col_shift, y_pos), 1, 1, color=xkcd))
  ax.text(0 + col_shift, y_pos + 0.5, " " + css4, alpha=0.5, **text_args)
  ax.text(1 + col_shift, y_pos + 0.5, " " + xkcd, alpha=0.5, **text_args)
  ax.text(2 + col_shift, y_pos + 0.5, " " + color_name, **text_args)

for g in range(n_groups):
  ax.hlines(range(n_rows), 3 * g, 3 * g + 2.8, color="0.7", linewidth=1)
  ax.text(0.5 + 3 * g, -0.5, "X11", ha="center", va="center")
  ax.text(1.5 + 3 * g, -0.5, "xkcd", ha="center", va="center")

ax.set_xlim(0, 3 * n_groups)
ax.set_ylim(n_rows, -1)
ax.axis("off")
plt.show()
````

### Latihan 3: Styling dengan Cycler

Styling dengan Cycler dalam Python, khususnya pada pustaka Matplotlib, adalah teknik untuk mengatur siklus properti visual grafik secara otomatis. Properti tersebut dapat berupa warna (color), jenis garis (linestyle), ketebalan garis (linewidth), dan bentuk penanda (marker).

Dengan Cycler, kita tidak perlu menentukan warna atau gaya setiap garis secara manual karena Matplotlib akan menggunakan properti secara berurutan.

````
from cycler import cycler
import matplotlib.pyplot as plt
import numpy as np

x = np.linspace(0, 2 * np.pi, 50)
offsets = np.linspace(0, 2 * np.pi, 4, endpoint=False)
yy = np.transpose([np.sin(x + phi) for phi in offsets])

default_cycler = cycler(color=["r", "g", "b", "y"]) + cycler(
    linestyle=["-", "--", ":", "-."]
)
plt.rc("lines", linewidth=4)
plt.rc("axes", prop_cycle=default_cycler)

custom_cycler = cycler(color=["c", "m", "y", "k"]) + cycler(
    lw=[1, 2, 3, 4]
)
fig, (ax0, ax1) = plt.subplots(nrows=2, figsize=(8, 8))
ax0.plot(yy)
ax0.set_title("Set default color cycle to rgby")
ax1.set_prop_cycle(custom_cycler)
ax1.plot(yy)
ax1.set_title("Set axes color cycle to cmyk")
fig.subplots_adjust(hspace=0.3)
plt.show()
````

### Latihan 4: Basic Continuous Colorbar

Basic Continuous Colorbar pada Python, khususnya menggunakan pustaka `Matplotlib`, adalah teknik untuk menampilkan skala warna kontinu yang merepresentasikan nilai numerik pada visualisasi data.

Colorbar membantu pembaca memahami hubungan antara warna dan nilai data. Contohnya, pada peta suhu, warna biru dapat menunjukkan suhu rendah, sedangkan warna merah menunjukkan suhu tinggi.

````
import matplotlib as mpl
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 1))
fig.subplots_adjust(bottom=0.5)
cmap = mpl.cm.cool
norm = mpl.colors.Normalize(vmin=5, vmax=10)
fig.colorbar(
    mpl.cm.ScalarMappable(norm=norm, cmap=cmap),
    cax=ax,
    orientation="horizontal",
    label="Some Units",
)
plt.show()
````

# Manipulasi Kanal Warna RGB & Operasi Piksel

Materi ini membahas cara memanipulasi kanal warna RGB (Red, Green, Blue) pada citra digital menggunakan Python. Dengan pustaka `Pillow`, `NumPy`, dan `Matplotlib`, kita dapat memisahkan kanal warna, melakukan operasi pada piksel, dan menampilkan hasil pengolahan citra.

## 1. Konsep dasar RGB dan operasi piksel

Citra digital berwarna umumnya direpresentasikan menggunakan tiga kanal warna:

## 3. Konsep operasi piksel

Operasi piksel adalah manipulasi nilai numerik pada satu atau banyak piksel citra. Pada kode di atas, operasi dilakukan dengan mengubah nilai kanal warna menjadi nol.

Misalkan satu piksel awal memiliki nilai:

\\[ P=(R,G,B)=(180,120,60) \\]

Maka hasil pemisahannya adalah:

- Kanal merah: \\(P_R=(180,0,0)\\)
- Kanal hijau: \\(P_G=(0,120,0)\\)
- Kanal biru: \\(P_B=(0,0,60)\\)

Perlu diperhatikan bahwa ketiga hasil ini bukan citra grayscale. Masing-masing merupakan citra RGB yang hanya mempertahankan satu kanal warna.


Kesimpulan: Manipulasi kanal RGB merupakan dasar penting dalam pengolahan citra digital. Dengan memahami struktur array dan operasi piksel, mahasiswa dapat mempelajari teknik lanjutan seperti konversi grayscale, penyesuaian intensitas, thresholding, segmentasi warna, dan ekstraksi fitur citra.

````
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

try:
    # Membuka gambar kucing.jpg dan mengubahnya menjadi array NumPy
    im = np.array(Image.open("orange.jpg"))

    # Memisahkan kanal Merah, Hijau, dan Biru

    # Menyalin gambar asli untuk pengolahan kanal merah
    im_R = im.copy()
    im_R[:, :, (1, 2)] = 0  # Menghilangkan kanal Hijau dan Biru

    # Menyalin gambar asli untuk pengolahan kanal hijau
    im_G = im.copy()
    im_G[:, :, (0, 2)] = 0  # Menghilangkan kanal Merah dan Biru

    # Menyalin gambar asli untuk pengolahan kanal biru
    im_B = im.copy()
    im_B[:, :, (0, 1)] = 0  # Menghilangkan kanal Merah dan Hijau

    # Menggabungkan ketiga gambar secara horizontal (axis=1)
    im_RGB = np.concatenate((im_R, im_G, im_B), axis=1)

    # Mengubah array NumPy menjadi objek gambar Pillow
    pil_img = Image.fromarray(im_RGB)

    # Menyimpan hasil pemisahan kanal ke file PNG
    pil_img.save("orange_channels_result.png")

    # Membuat area gambar berukuran 12 x 4 inci
    plt.figure(figsize=(12, 4))

    # Menampilkan gambar hasil penggabungan kanal
    plt.imshow(pil_img)

    # Memberikan judul pada visualisasi
    plt.title("Pemisahan Kanal RGB (Red | Green | Blue)")

    # Menyembunyikan sumbu koordinat agar gambar lebih jelas
    plt.axis("off")

    # Menampilkan hasil visualisasi
    plt.show()

except Exception as e:
    # Menampilkan pesan jika terjadi kesalahan saat memproses gambar
    print("Pastikan file 'orange.jpg' tersedia untuk menjalankan pemisahan kanal.")
````
