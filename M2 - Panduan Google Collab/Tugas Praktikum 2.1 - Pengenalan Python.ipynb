## 3. String Variables


| Kode           | Fungsi                                                                         |
| -------------- | ------------------------------------------------------------------------------ |
| `upper()`      | Mengubah semua huruf menjadi kapital.                                          |
| `lower()`      | Mengubah semua huruf menjadi kecil.                                            |
| `capitalize()` | Membuat huruf pertama kalimat menjadi kapital dan huruf lainnya menjadi kecil. |
| `title()`      | Membuat huruf pertama setiap kata menjadi kapital.                             |
| `count('a')`   | Menghitung jumlah kemunculan huruf `a`.                                        |
| `+`            | Menggabungkan dua atau lebih string.                                           |
| `' '`          | Memberikan spasi saat menggabungkan string.                                    |

Python membedakan huruf besar dan kecil. Misalnya, `count('a')` tidak menghitung huruf `A` kapital.

```python

# 1. Menyimpan teks dalam variabel
nama_depan = 'Andi'
nama_belakang = 'Pratama'

print('Nama depan:', nama_depan)
print('Nama belakang:', nama_belakang)

print()

# 2. Mengubah format huruf pada string
kalimat = 'saya sedang belajar pemrograman python'

print('Kalimat asli:', kalimat)
print('Huruf kapital semua:', kalimat.upper())
print('Huruf kecil semua:', kalimat.lower())
print('Huruf awal kapital:', kalimat.capitalize())
print('Setiap Kata Diawali Kapital:', kalimat.title())

print()

# 3. Menghitung jumlah karakter tertentu
teks = 'belajar bahasa python'

print('Teks:', teks)
print("Jumlah huruf 'a':", teks.count('a'))
print("Jumlah huruf 'b':", teks.count('b'))
print("Jumlah huruf 'o':", teks.count('o'))

print()

# 4. Menggabungkan dua string menggunakan operator +
nama_depan = 'Siti'
nama_belakang = 'Aisyah'

# Menggabungkan tanpa spasi
print('Nama tanpa spasi:', nama_depan + nama_belakang)

# Menggabungkan dengan spasi
print('Nama lengkap:', nama_depan + ' ' + nama_belakang)

print()

# 5. Menggabungkan teks dan variabel
program_studi = 'Informatika'

print('Halo, nama saya ' + nama_depan + ' ' + nama_belakang)
print('Saya mahasiswa program studi ' + program_studi)
print('Saya sedang belajar ' + 'Python')
```

## 4. Numerical Variables



## Penjelasan singkat

| Kode      | Fungsi                                      |
| --------- | ------------------------------------------- |
| `int()`   | Mengubah input menjadi bilangan bulat.      |
| `float()` | Mengubah input menjadi bilangan desimal.    |
| `+`       | Penjumlahan.                                |
| `-`       | Pengurangan.                                |
| `*`       | Perkalian.                                  |
| `/`       | Pembagian biasa.                            |
| `**`      | Perpangkatan.                               |
| `//`      | Pembagian dengan hasil dibulatkan ke bawah. |
| `%`       | Menghasilkan sisa pembagian.                |
| `input()` | Menerima masukan dari pengguna.             |

Hasil `input()` selalu berupa string. Karena itu, gunakan `int()` atau `float()` agar input dapat dioperasikan sebagai angka.

Contoh: jika pengguna memasukkan `10` dan `3`, hasil penjumlahan adalah `13`, sedangkan hasil pembagian biasa adalah `3.3333333333333335`.

```python

# 1. Menyimpan nilai numerik dalam variabel
harga_barang = 15000
print('Harga barang:', harga_barang)

# Menyimpan bilangan desimal
nilai_pi = 3.14159
print('Nilai PI:', nilai_pi)

print('-----------------------------------')

# 2. Melakukan operasi matematika
angka_pertama = 10
angka_kedua = 3

print('Angka pertama:', angka_pertama)
print('Angka kedua:', angka_kedua)

print('Penjumlahan:', angka_pertama + angka_kedua)
print('Pengurangan:', angka_pertama - angka_kedua)
print('Perkalian:', angka_pertama * angka_kedua)
print('Pembagian:', angka_pertama / angka_kedua)
print('Pangkat:', angka_pertama ** angka_kedua)
print('Pembagian bulat:', angka_pertama // angka_kedua)
print('Sisa pembagian:', angka_pertama % angka_kedua)

print('-----------------------------------')

# 3. Menerima input angka dari pengguna
angka1 = input('Masukkan angka pertama: ')
angka2 = input('Masukkan angka kedua: ')

# Mengubah input menjadi integer
hasil_tambah = int(angka1) + int(angka2)
print('Hasil penjumlahan integer:', hasil_tambah)

# Mengubah input menjadi float
hasil_desimal = float(angka1) + float(angka2)
print('Hasil penjumlahan float:', hasil_desimal)

print('-----------------------------------')

# 4. Contoh menghitung luas persegi panjang
panjang = float(input('Masukkan panjang: '))
lebar = float(input('Masukkan lebar: '))

luas = panjang * lebar

print('Panjang:', panjang)
print('Lebar:', lebar)
print('Luas persegi panjang:', luas)
```

## 5. Loops


## Penjelasan singkat

| Kode                              | Fungsi                                                         |
| --------------------------------- | -------------------------------------------------------------- |
| `for nama in ...`                 | Mengulang setiap nama dalam daftar.                            |
| `range(1, 6)`                     | Menghasilkan angka 1 sampai 5.                                 |
| `while indeks < len(daftar_nama)` | Mengulang selama indeks masih berada dalam batas daftar.       |
| `while angka in range(1, 6)`      | Mengulang selama angka termasuk dalam rentang 1–5.             |
| `indeks = indeks + 1`             | Menambah indeks agar perulangan bergerak ke elemen berikutnya. |
| `total = total + nilai`           | Menjumlahkan seluruh nilai dalam daftar.                       |
| `len()`                           | Menghitung jumlah elemen dalam daftar.                         |

Pada perulangan `while`, variabel kondisi harus diperbarui agar perulangan dapat berhenti. Jika nilai `indeks` atau `angka` tidak berubah, program dapat mengalami perulangan tanpa akhir.

```python

# 1. Perulangan for dengan variabel string
print('=== Perulangan for dengan Nama Mahasiswa ===')

for nama in ['Andi', 'Budi', 'Siti', 'Rina']:
    print(nama)

print()

# 2. Perulangan for dengan variabel numerik
print('=== Perulangan for dengan Angka ===')

for angka in range(1, 6):
    print('Angka ke-', angka)

print()

# 3. Perulangan while dengan variabel string
print('=== Perulangan while dengan Nama Mahasiswa ===')

daftar_nama = ['Andi', 'Budi', 'Siti', 'Rina']
indeks = 0

while indeks < len(daftar_nama):
    print(daftar_nama[indeks])
    indeks = indeks + 1

print()

# 4. Perulangan while dengan variabel numerik
print('=== Perulangan while dengan Angka ===')

angka = 1

while angka in range(1, 6):
    print('Nilai angka:', angka)
    angka = angka + 1

print()

# 5. Perulangan for untuk menghitung total angka
print('=== Menghitung Total Nilai ===')

nilai_mahasiswa = [80, 75, 90, 85]
total = 0

for nilai in nilai_mahasiswa:
    print('Nilai:', nilai)
    total = total + nilai

print('Total nilai:', total)
print('Rata-rata nilai:', total / len(nilai_mahasiswa))
```

## 6. Function


## Penjelasan kode

- `import datetime` — mengimpor modul untuk mengakses tanggal dan waktu.
- `def catat_waktu():` — membuat fungsi bernama `catat_waktu`.
- `datetime.datetime.now()` — mengambil tanggal dan waktu saat program dijalankan.
- `catat_waktu()` — memanggil fungsi agar instruksi di dalamnya dijalankan.
- `for nomor in range(1, 6):` — mengulang proses dari nomor 1 sampai 5.
- `for tugas in daftar_tugas:` — menampilkan setiap tugas dalam daftar.
- `'-' * 40` — mencetak karakter garis sebanyak 40 kali sebagai pemisah.

 Fungsi dapat dipanggil berkali-kali tanpa perlu menulis ulang seluruh instruksi di dalamnya. Pada contoh ini, fungsi `catat_waktu()` dipanggil sebelum dan sesudah proses tertentu untuk mencatat waktu penyelesaiannya.

```python

import datetime

# 1. Membuat fungsi untuk menampilkan waktu
def catat_waktu():
    print('Tugas berhasil diselesaikan!')
    print('Waktu:', datetime.datetime.now())
    print('-' * 40)

# 2. Menyimpan data mahasiswa
nama = 'Andi'
mata_kuliah = 'Struktur Data'

print('=== INFORMASI MAHASISWA ===')
print('Nama:', nama)
print('Mata Kuliah:', mata_kuliah)

# Memanggil fungsi untuk mencatat waktu
catat_waktu()

# 3. Melakukan perulangan angka
print('=== PROSES PENGERJAAN TUGAS ===')

for nomor in range(1, 6):
    print('Mengerjakan soal nomor', nomor)

# Memanggil kembali fungsi setelah perulangan selesai
catat_waktu()

# 4. Melakukan perulangan daftar tugas
print('=== DAFTAR TUGAS ===')

daftar_tugas = [
    'Mempelajari variabel',
    'Mempelajari fungsi',
    'Mempelajari perulangan'
]

for tugas in daftar_tugas:
    print('-', tugas)

# Mencatat waktu setelah semua tugas ditampilkan
catat_waktu()
```
