## C. Daftar tugas praktikum

**Tugas 1 — Menampilkan Informasi dengan `print()`**

Buat program untuk menampilkan biodata mahasiswa dengan ketentuan:

- Tampilkan judul `BIODATA MAHASISWA`.
- Tampilkan nama, NIM, program studi, dan universitas.
- Gunakan petik tunggal dan petik ganda.
- Gunakan `print()` kosong untuk membuat baris kosong.
- Gunakan `\n` untuk menampilkan beberapa baris dari satu perintah `print()`.
- Gunakan karakter `=` atau `-` sebagai garis pemisah.

Output yang diharapkan: biodata mahasiswa tersusun rapi di terminal.

**Tugas 2 — Mengolah String**

Buat program untuk mengolah data nama dan kalimat dengan ketentuan:

- Simpan nama depan dan nama belakang dalam variabel.
- Gabungkan keduanya menjadi nama lengkap.
- Tampilkan sebuah kalimat menggunakan `upper()`, `lower()`, `capitalize()`, dan `title()`.
- Hitung jumlah kemunculan huruf `a` menggunakan `count()`.
- Tampilkan hasil setiap operasi.

Output yang diharapkan: nama lengkap, perubahan format kalimat, dan jumlah karakter.



**Tugas 3 — Operasi Matematika**

Buat program kalkulator sederhana dengan ketentuan:

- Minta pengguna memasukkan dua angka.
- Konversikan input menjadi `int` atau `float`.
- Hitung penjumlahan, pengurangan, perkalian, pembagian, dan pangkat.
- Tampilkan hasil setiap operasi.
- Tambahkan perhitungan luas persegi panjang berdasarkan input panjang dan lebar.

Output yang diharapkan: hasil kalkulasi berdasarkan angka yang dimasukkan pengguna.

Tugas 4 — Perulangan `for` dan `while`

Buat program untuk menampilkan data mahasiswa dengan ketentuan:

- Buat daftar minimal lima nama mahasiswa.
- Tampilkan semua nama menggunakan `for`.
- Tampilkan angka 1 sampai 10 menggunakan `range()`.
- Tampilkan kembali daftar nama menggunakan `while` dan `len()`.
- Buat perulangan `while` untuk menampilkan bilangan genap dari 2 sampai 20.

Output yang diharapkan: daftar nama dan deret angka yang tampil sesuai ketentuan.

Tugas 5 — Fungsi dan Waktu

Buat program pencatatan aktivitas belajar dengan ketentuan:

- Impor modul `datetime`.
- Buat fungsi bernama `catat_waktu()`.
- Tampilkan pesan `Aktivitas selesai`.
- Tampilkan waktu saat fungsi dijalankan menggunakan `datetime.datetime.now()`.
- Panggil fungsi sebelum dan sesudah perulangan.
- Gunakan perulangan `for` untuk menampilkan proses pengerjaan lima soal.

Output yang diharapkan: pesan aktivitas, waktu pelaksanaan, dan proses pengerjaan soal.

Studi kasus
